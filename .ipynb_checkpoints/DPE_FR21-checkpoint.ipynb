```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, HTML
import plotly.express as px
import geopandas as gpd
```

Data source: https://data.ademe.fr/datasets/dpe-v2-logements-neufs \
GES: greenhouse gases\
DPE: Energy Performance Diagnosis

# Data preparation

```python
dpe_av = pd.read_csv('data/dpe_before21.csv')
dpe_av.head()
```

```python
# Remove duplicates
dpe_av.drop_duplicates(keep='first',inplace=True)
```

```python
# missing values
(dpe_av.isna().sum()*100/len(dpe_av)).sort_values(ascending=False)
```

```python
# Delete useless columns
dpe_av.drop(columns = ['nom_methode_dpe','version_methode_dpe',
                       'latitude','longitude','date_etablissement_dpe','geo_adresse'],axis = 1, inplace =True)
```

```python
# Rename columns
dpe_av.rename(columns={'code_insee_commune_actualise' : 'code_commune','tv016_departement_code' : 'code_dep',\
                       'tr002_type_batiment_description' : 'description','surface_thermique_lot' : 'surface',\
                       'tr001_modele_dpe_type_libelle' : 'contrat','annee_construction':'annee',\
                       'tr001_modele_dpe_type_libelle' : 'vente/loca',\
                       'tr002_type_batiment_description' : 'type_bat'}, inplace=True)
```

```python
# Removal of lines without thermal surface. 9 rows
dpe_av.drop(dpe_av[dpe_av['surface'].isna()].index,inplace=True)

# Removal of lines without code_commune 94 lines
# We could also have assigned a code_commune from the code_dep
dpe_av.drop(dpe_av[dpe_av['code_commune'].isna()].index,inplace=True)
```

```python
# Remove outliers
dpe_av['annee'] = dpe_av['annee'].map(lambda x: 0 if x > 2021 or x <1800 else x)

# Replace 0 by nan
dpe_av[['consommation_energie','estimation_ges','annee']]= \
            dpe_av[['consommation_energie','estimation_ges','annee']].replace(0,np.nan)
```

```python
(dpe_av.isna().sum()*100/len(dpe_av)).sort_values(ascending=False)
```

```python
# Removal of missing construction year.
dpe_av = dpe_av.drop(dpe_av[dpe_av['annee'].isna()].index) #2549

# We remove the N classes that correspond to ungraded ones and will distort our calculations. about 15% of rows
dpe_av = dpe_av.drop(dpe_av[dpe_av['classe_consommation_energie'] == 'N'].index) 
```

```python
dpe_av[dpe_av['estimation_ges'].isna()]['classe_estimation_ges'].value_counts()
```

```python
# We replace the missing estimates with the average of category A and G.
# Calculate mean estimation_ges values for each classe_estimation_ges category
mean_values = dpe_av.groupby('classe_estimation_ges')['estimation_ges'].transform('mean')

# Assign mean values to missing estimation_ges based on their classe_estimation_ges category
dpe_av['estimation_ges'] = dpe_av['estimation_ges'].fillna(mean_values)
```

# Etude sur les DPE

```python
def func(pct, allvals):
    absolute = int(np.round(pct/np.sum(allvals)))
    return "{:.1f}%".format(pct, absolute)

x = dpe_av['classe_estimation_ges'].value_counts(normalize=True).sort_index()
myexplode = [0,0,0.2,0,0,0,0]

plt.pie(x,labels=['A','B','C','D','E','F','G'],explode = myexplode,shadow = True,\
       autopct=lambda pct: func(pct, x),textprops=dict(color="w"))

plt.legend(bbox_to_anchor=(1, 1))
plt.title("Distribution according to GHG production class estimates")
plt.show()
```

We can see from the pie chart above that class C is the most widespread class (22.3%) of greenhouse gas production.

```python
y = dpe_av['classe_consommation_energie'].value_counts(normalize=True).sort_index()
myexplode2 = [0,0,0,0.1,0,0,0]
plt.pie(y,labels=['A','B','C','D','E','F','G'],explode = myexplode2,shadow = True,\
       autopct=lambda pct: func(pct, x),textprops=dict(color="w"))
plt.legend(bbox_to_anchor=(1, 1))
plt.title("Distribution of energy consumption classes")
plt.show()
```

Regarding energy consumption, class D is the most widespread.

# Thermal strainers
Energy class is F or G

```python
passoire = dpe_av[dpe_av['classe_consommation_energie'].isin(['F','G'])]
```

```python
# Histogram distribution year of construction
plt.hist(dpe_av['annee'],30)
plt.title('Evolution of housing construction over time')
plt.show()
```

```python
plt.hist(passoire['annee'],30)
plt.title('Evolution of colander construction over time')
plt.show()
```

```python
print('Passoire',len(passoire[passoire['annee']<1950])/len(passoire))
print('Tout',len(dpe_av[dpe_av['annee']<1950])/len(dpe_av))
```

Both histograms follow the same trend, with a peak in construction in the post-war period, at the end of the 1940s.\
37% of the sieves were built before 1950, compared to only 30% of dwellings of all classes combined.

```python
fig, ax = plt.subplots()

x = dpe_av['annee'].sort_values(ascending=True).unique()
y = pd.Series(x.copy())
y = y.apply(lambda x:len(passoire[passoire['annee']<=x])/len(dpe_av[dpe_av['annee']<=x]))
ymax = max(y)
xpos = y.idxmax()
xmax = x[xpos]
ax.plot(x,y)
a1 = ax.annotate('{:.2f}'.format(ymax), xy=(xmax,ymax), xytext=(2000,ymax),
            arrowprops=dict(facecolor='black', shrink=0.05))

a2 = ax.annotate('{:.2f}'.format(y[0]), xy=(x[0],y[0]), xytext=(x[0],y[0]-0.008))

a3 = ax.annotate('{:.2f}'.format(y[114]), xy=(x[114],y[114]), xytext=(x[120],y[114]))

plt.title('Percentage of thermal strainer over time')
plt.xlabel('year')
plt.ylabel('% of thermail strainer')
plt.show()
```

The percentage of colanders in France was slightly decreasing (21% ->20%) until the end of the war before experiencing a sharp explosion over 25 years to reach 27% and then a sharp decline (21%). Overall, the percentage of colanders is 20%.

```python
x = dpe_av['consommation_energie'].mean()
bins = np.linspace(0, 500, 25)

plt.hist(dpe_av[dpe_av['classe_consommation_energie']=='A']['consommation_energie'],bins,alpha=0.8, label='A')
plt.hist(dpe_av[dpe_av['classe_consommation_energie']=='B']['consommation_energie'],bins,alpha=0.8, label='B')
plt.hist(dpe_av[dpe_av['classe_consommation_energie']=='C']['consommation_energie'],bins,alpha=0.8, label='C')
plt.hist(dpe_av[dpe_av['classe_consommation_energie']=='D']['consommation_energie'],bins,alpha=0.8, label='D')
plt.hist(dpe_av[dpe_av['classe_consommation_energie']=='E']['consommation_energie'],bins,alpha=0.8, label='E')
plt.hist(dpe_av[dpe_av['classe_consommation_energie']=='F']['consommation_energie'],bins,alpha=0.8, label='F')
plt.hist(dpe_av[dpe_av['classe_consommation_energie']=='G']['consommation_energie'],bins,alpha=0.8, label='G')

min_ylim, max_ylim = plt.ylim()
plt.axvline(x, color='k', linestyle='dashed', linewidth=1)
plt.text(x*1.1, max_ylim*0.9, 'Mean: {:.2f}'.format(x))
plt.legend(loc='upper right')
plt.title("energy consumption by class")
plt.xlabel('consommation_energie')
plt.show()
```

We note that the different energy consumption classes are well delimited. Although the average (252) belongs to E, it is class C which is the most represented on this histogram.

```python
x = dpe_av['estimation_ges'].mean()
bins = np.linspace(0, 125, 25)

plt.hist(dpe_av[dpe_av['classe_estimation_ges']=='A']['estimation_ges'],bins,alpha=0.8, label='A')
plt.hist(dpe_av[dpe_av['classe_estimation_ges']=='B']['estimation_ges'],bins,alpha=0.8, label='B')
plt.hist(dpe_av[dpe_av['classe_estimation_ges']=='C']['estimation_ges'],bins,alpha=0.8, label='C')
plt.hist(dpe_av[dpe_av['classe_estimation_ges']=='D']['estimation_ges'],bins,alpha=0.8, label='D')
plt.hist(dpe_av[dpe_av['classe_estimation_ges']=='E']['estimation_ges'],bins,alpha=0.8, label='E')
plt.hist(dpe_av[dpe_av['classe_estimation_ges']=='F']['estimation_ges'],bins,alpha=0.8, label='F')
plt.hist(dpe_av[dpe_av['classe_estimation_ges']=='G']['estimation_ges'],bins,alpha=0.8, label='G')

min_ylim, max_ylim = plt.ylim()
plt.axvline(x, color='k', linestyle='dashed', linewidth=1)
plt.text(x*1.1, max_ylim*0.9, 'Mean: {:.2f}'.format(x))
plt.legend(loc='upper right')
plt.title("Estimation of GHG production based on GHG class")
plt.xlabel('estimation_ges')
plt.show()
```

The average estimate of greenhouse gas production is 33. On average, the dwellings in this sample are therefore class D.

```python
plt.hist(passoire['surface'],bins,alpha=1, label='Passoire',stacked =True,color='b')
plt.hist(dpe_av['surface'],bins,alpha=0.5, label='Moyenne')
plt.legend(loc='upper right')
plt.title('Housing area depending on type')
plt.xlabel('surface area')
plt.show()
```

The histogram of the colander is similar (to a lesser extent) to that of the average of the dwellings. There is therefore no link between the surface area of ​​a dwelling and whether it is a colander or not.

```python
# Number of colanders per department 

pass_dep = pd.DataFrame(passoire.groupby(['code_dep']).count().iloc[:,5])
b = dpe_av.groupby(['code_dep']).count().iloc[:,5]
pass_dep = pass_dep.rename(columns={'surface' : 'nb_passoire'})
pass_dep['pourcentage'] = pass_dep['nb_passoire']/b
pass_dep.reset_index(inplace=True)

pd.set_option("display.max_rows", None)
display(HTML("<div style='height: 200px; overflow: auto; width: fit-content'>" +
             pass_dep.style.to_html() +
             "</div>"))

pd.reset_option("display.max_rows")
```

```python
# We import a file allowing the division of regions in France
geo = gpd.read_file('data/a-dep2021.json')
geo.head()
```

```python
# We retrieve the name of the department
pass_dep = pd.merge(pass_dep,geo, left_on='code_dep', right_on='dep', how='left')
pass_dep = pass_dep[['code_dep','nb_passoire','pourcentage','libgeo']]
```

The following map allows us to visualize the percentage of colanders in France.

```python
fig = px.choropleth_mapbox(pass_dep,
                           locations='code_dep',
                           geojson=geo,
                           color='pourcentage',
                           featureidkey="properties.dep",
                           mapbox_style='carto-positron',
                           hover_name = 'libgeo',
                           hover_data = ['code_dep','nb_passoire'],
                           center={'lat': 47.7516, 'lon': 1.6751},
                           zoom=4,
                           title = 'pourcentage de passoire thermique en France'
                          )
fig.update_layout(margin={'l': 10, 'r': 10, 't': 10, 'b': 10})
fig.show()
```
